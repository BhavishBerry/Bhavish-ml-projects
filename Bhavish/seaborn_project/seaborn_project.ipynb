```python
## BOXPLOT - performance score spread per sport
plt.figure(figsize=(10, 6))
sns.boxplot(x='sport', y='performance_score', data=df)
plt.title('Performance Score by Sport', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
## SCATTERPLOT - training hours vs performance score
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='training_hours_per_week', y='performance_score', hue='sport')
plt.title('Training Hours vs Performance Score', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
## VIOLINPLOT - vo2 max distribution per sport
plt.figure(figsize=(10, 6))
sns.violinplot(x='sport', y='vo2_max', data=df)
plt.title('VO2 Max Distribution by Sport', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
## BARPLOT - average resting heart rate per sport
plt.figure(figsize=(10, 6))
sns.barplot(x='sport', y='resting_heart_rate', data=df, hue='sport', palette='Blues_r', legend=False)
plt.title('Average Resting Heart Rate by Sport', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
## HEATMAP - correlation between numeric features
plt.figure(figsize=(10, 8))
numeric_df = df.drop(columns=['athlete_id', 'sport'])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
## PAIRPLOT - relationships across key numeric features
sns.pairplot(df[['training_hours_per_week', 'vo2_max', 'performance_score', 'sport']], hue='sport')
plt.show()
```
